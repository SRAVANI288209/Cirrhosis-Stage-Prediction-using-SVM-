```python
# Outlier analysis using IQR

# Identify numerical columns automatically
num_features = df.select_dtypes(include=['number']).columns.tolist()

# Remove Stage because it is the target column
num_features = [col for col in num_features if col != 'Stage']

outlier_summary = []

for col in num_features:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    
    iqr = q3 - q1
    
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()
    
    outlier_summary.append({
        'Feature': col,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': count
    })

# Convert results into DataFrame
outlier_summary_df = pd.DataFrame(outlier_summary)

# Sort by number of outliers
outlier_summary_df = outlier_summary_df.sort_values(
    'Outlier Count',
    ascending=False
)

# Display result
outlier_summary_df
```

```python
#outlier analysis using the IQR method identifies potential outliers in numerical features. The summary includes Q1, Q3, IQR, lower and upper bounds, and the count of outliers for each feature. This information helps in understanding the distribution of data and deciding on strategies for handling outliers during preprocessing.
```

```python
# ============================================================
# OUTLIER HANDLING USING IQR CAPPING (WINSORIZATION)
# ============================================================

# Selected continuous/numerical features
cap_columns = [
    'Bilirubin',
    'Cholesterol',
    'Copper',
    'Alk_Phos',
    'Tryglicerides',
    'Platelets',
    'Albumin',
    'SGOT',
    'Prothrombin',
    'Age'
]

# ------------------------------------------------------------
# Step 1: Calculate IQR bounds for each selected column
# ------------------------------------------------------------

cap_summary = []

for col in cap_columns:

    # First quartile
    q1 = df[col].quantile(0.25)

    # Third quartile
    q3 = df[col].quantile(0.75)

    # Interquartile range
    iqr = q3 - q1

    # Lower and upper limits
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Store the results
    cap_summary.append({
        'Feature': col,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound
    })

# Create summary DataFrame
cap_summary_df = pd.DataFrame(cap_summary)

# Display the IQR bounds
print("IQR CAPPING SUMMARY")
display(cap_summary_df)


# ------------------------------------------------------------
# Step 2: Apply IQR capping
# ------------------------------------------------------------

for col in cap_columns:

    # Calculate Q1 and Q3
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)

    # Calculate IQR
    iqr = q3 - q1

    # Calculate limits
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Cap values outside the limits
    df[col] = df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )


# ------------------------------------------------------------
# Step 3: Display confirmation
# ------------------------------------------------------------

print("\nOutlier capping applied successfully.")

print("\nColumns where capping was applied:")
print(cap_columns)


# ------------------------------------------------------------
# Step 4: Display sample data after capping
# ------------------------------------------------------------

print("\nSample data after capping:")

display(
    df[
        ['Bilirubin',
         'Cholesterol',
         'Copper',
         'Alk_Phos',
         'Tryglicerides',
         'Platelets',
         'Albumin',
         'SGOT',
         'Prothrombin',
         'Age']
    ].head()
)


# ------------------------------------------------------------
# Step 5: Check whether the code worked
# ------------------------------------------------------------

print("\nDataset shape after capping:")
print(df.shape)

print("\nMissing values in capped columns:")
print(df[cap_columns].isnull().sum())
```

```python
#handling outliers using IQR capping (winsorization) is a technique to limit extreme values in numerical features. The code calculates the IQR bounds for selected continuous features and applies capping to values outside these bounds. This helps in reducing the influence of outliers on the model while preserving the overall distribution of the data. The summary of IQR bounds, confirmation of capping, and sample data after capping are displayed to ensure that the operation was successful.
```

```python
# ============================================================
# CORRELATION ANALYSIS FOR NUMERICAL VARIABLES
# ============================================================

# Select only numerical columns
# Exclude Stage because it is the target variable
num_corr_df = df.select_dtypes(include=['number']).drop(
    columns=['Stage'],
    errors='ignore'
)

# ============================================================
# STEP 1: CREATE CORRELATION MATRIX
# ============================================================

correlation_matrix = num_corr_df.corr()

print("Correlation Matrix:")
display(correlation_matrix)


# ============================================================
# STEP 2: CREATE CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=False,
    cmap='coolwarm',
    center=0
)

plt.title('Numerical Feature Correlation Heatmap')
plt.tight_layout()
plt.show()


# ============================================================
# STEP 3: FIND STRONG CORRELATIONS
# ============================================================

# Get only the upper triangle of the correlation matrix
upper_triangle = np.triu(
    np.ones(correlation_matrix.shape),
    k=1
).astype(bool)

strong_pairs = (
    correlation_matrix
    .where(upper_triangle)
    .stack()
    .sort_values(ascending=False)
)

# Keep correlations where absolute correlation is greater than 0.3
strong_pairs = strong_pairs[strong_pairs.abs() > 0.3]


# ============================================================
# STEP 4: DISPLAY STRONG CORRELATIONS
# ============================================================

print("Strong feature correlations (|r| > 0.3):")

if len(strong_pairs) > 0:
    print(strong_pairs)
else:
    print("No strong correlations found.")
```

```python

# Step 15: preprocessing pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

X = df.drop(columns=['Stage', 'ID'])
y = df['Stage']

numeric_features = X.select_dtypes(include=['number']).columns.tolist()
categorical_features = X.select_dtypes(exclude=['number']).columns.tolist()

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print(f'Original X_train shape: {X_train.shape}')
print(f'Processed X_train shape: {X_train_processed.shape}')
print(f'Processed X_test shape: {X_test_processed.shape}')
print(f'Numeric features: {numeric_features}')
print(f'Categorical features: {categorical_features}')
```

```python

# Step 16: apply SMOTE only to training data after preprocessing
from imblearn.over_sampling import SMOTE

X = df.drop(columns=['Stage', 'ID'])
y = df['Stage']

numeric_features = X.select_dtypes(include=['number']).columns.tolist()
categorical_features = X.select_dtypes(exclude=['number']).columns.tolist()

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print('Before SMOTE class distribution (training):')
print(y_train.value_counts().sort_index())

smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_processed, y_train)

print('\nAfter SMOTE class distribution (training):')
print(pd.Series(y_train_resampled).value_counts().sort_index())

print(f'\nOriginal training shape: {X_train_processed.shape}')
print(f'Resampled training shape: {X_train_resampled.shape}')
print(f'Test shape: {X_test_processed.shape}')
```

```python
####The output shows that **SMOTE (Synthetic Minority Over-sampling Technique)** is applied only to the **training data after preprocessing**. It creates new synthetic samples for the minority classes so that the class distribution becomes more balanced. The testing data is kept unchanged because it should represent the original data distribution for a fair evaluation of the model. This helps the machine learning model learn the minority classes more effectively and reduces the effect of class imbalance during training.
```
